```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_absolute_error
import joblib
```

```python
df = pd.read_csv('training-data.csv')
print(f"Total rows: {len(df)}")
print(df.head())
print(df.describe())
```

# SPLITTING FEATURES AND TARGETS

```python
features = ['level', 'temperature', 'tds', 'ph', 'turbidity']
X = df[features]

y_freshness = df['freshness_window_hours']
y_risk = df['risk_score']

X_train, X_test, y_freshness_train, y_freshness_test = train_test_split(
    X, y_freshness, test_size=0.2, random_state=42
)

_, _, y_risk_train, y_risk_test = train_test_split(
    X, y_risk, test_size=0.2, random_state=42
)

print(f"Training samples: {len(X_train)}")
print(f"Testing samples: {len(X_test)}")
```

# TRAINING TWO SEPERATE RANDOM FOREST MODELS

```python
freshness_model = RandomForestRegressor(n_estimators=100, max_depth=10, random_state=42)
freshness_model.fit(X_train, y_freshness_train)

risk_model = RandomForestRegressor(n_estimators=100, max_depth=10, random_state=42)
risk_model.fit(X_train, y_risk_train)

print("Both models trained successfully")
```

# EVALUATING MODEL PERFORMANCE

```python
freshness_pred = freshness_model.predict(X_test)
risk_pred = risk_model.predict(X_test)

print("Freshness Window Model:")
print(f"  R² Score: {r2_score(y_freshness_test, freshness_pred):.4f}")
print(f"  Mean Absolute Error: {mean_absolute_error(y_freshness_test, freshness_pred):.4f} hours")

print("\nRisk Score Model:")
print(f"  R² Score: {r2_score(y_risk_test, risk_pred):.4f}")
print(f"  Mean Absolute Error: {mean_absolute_error(y_risk_test, risk_pred):.4f}")
```

# EXPORTING THE MODELS AS PKL FILES

```python
joblib.dump(freshness_model, 'aquasense_freshness_model.pkl')
joblib.dump(risk_model, 'aquasense_risk_model.pkl')

files.download('aquasense_freshness_model.pkl')
files.download('aquasense_risk_model.pkl')

print("Models exported and downloaded")
```
